```python
import pandas as pd
import numpy as np
```

```python
schools = "Resources/schools_complete.csv"
students = "Resources/students_complete.csv"
```

```python
school_data = pd.read_csv(schools)
student_data = pd.read_csv(students)
```

```python
school_data.head()
#student_data.head()
```

```python
combined_data = pd.merge(school_data, student_data, on = ["school_name", "school_name"])
#combined_data_df = pd.DataFrame(combined_data)
combined_data.head()
```

```python
global_avg_math_score = round((combined_data['math_score'].sum())/(len(combined_data['student_name'])),3)
global_avg_reading_score = round((combined_data['reading_score'].sum())/(len(combined_data['student_name'])),3)

print(f'The average math score for all students was {global_avg_math_score}.')
print(f'The average reading score for all students was {global_avg_reading_score}.')
```

```python
total_school = len(combined_data['school_name'].unique())
total_students = len(combined_data['student_name'])
print(f'There are {total_school} schools with a total of {total_students} students.')
total_budget = school_data['budget'].sum()
print(total_budget)
```

```python
passing_math_count = combined_data[(combined_data["math_score"] >= 70)].count()["student_name"]
passing_math_percentage = round(passing_math_count / float(total_students) * 100,3)
passing_reading_count = combined_data[(combined_data["reading_score"] >= 70)].count()["student_name"]
passing_reading_percentage = round(passing_reading_count / float(total_students) * 100,3)
```

```python
print(f'{passing_math_percentage}% and {passing_reading_percentage}%')
```

```python
District_overview = pd.DataFrame({"Total Schools": [total_school], "Total Students": [total_students], 
                                  "Percent passing math": [passing_math_percentage], "Percent passing reading":[passing_reading_percentage], "Total Budget": [total_budget] })
District_overview
```
