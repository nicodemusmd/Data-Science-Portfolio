```python
import pandas as pd
import numpy as np
```

```python
school_data_to_load = "Resources/schools_complete.csv"
student_data_to_load = "Resources/students_complete.csv"
```

```python
school_data = pd.read_csv(school_data_to_load)
student_data = pd.read_csv(student_data_to_load)
```

```python
school_data_complete = pd.merge(school_data, student_data, 
                                how = "left", on = ["school_name", "school_name"] )
```

```python
school_data_complete.head()
```

```python
school_num = len(school_data_complete["school_name"].unique())
student_num = len(school_data_complete["student_name"].unique())
print(f'There are a total of {school_num} schools and {student_num} students') 
```

```python
total_budget = school_data["budget"].sum()
print(f'The total budget for the school district is {total_budget}.')
```

```python
avg_math_score = student_data["math_score"].mean()
avg_reading_score = student_data["reading_score"].mean()
print(f'The average math score is {round(avg_math_score,3)} and the average reading score is {round(avg_reading_score, 3)}.')

avg_score = (avg_math_score + avg_reading_score)/2
print(f'The average total score is {round(avg_score,3)}.')
```
